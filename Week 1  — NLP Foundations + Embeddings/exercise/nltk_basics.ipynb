import the stop words and create word then tokenize


```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
st_word = 'Sir, I protest. I am not a merry man!'
word_in_quote = word_tokenize(st_word)
word_in_quote
```

to create a set of stop words to filter words_in_quote. For this example, you’ll need to focus on stop words in "english":


```python
stop_words = set(stopwords.words('english'))
# to iterat over words_in_quote with a for loop and added all the words that weren’t stop words to filtered_list
filtered_word = [word for word in word_in_quote if word.casefold() not in stop_words]
filtered_word
```

normal for loop for itration


```python
filtered_wor = []
for word in word_in_quote:
    if word.casefold() not in stop_words:
        filtered_wor.append(word)
filtered_wor
```

CHUNKING
  Process:
    1.tokenize
    2.pos Tag
    3.chunk

```python
import nltk
from nltk.tokenize import word_tokenize
gram  = "It's a dangerous business, Frodo, going out your door."
gram_tokenize = word_tokenize(gram)
gram_pos_tag = nltk.pos_tag(gram_tokenize)
# Define a simple grammar for noun phrases
grammar = "NP: {<DT>?<JJ>*<NN>}"
chunk_parser = nltk.RegexpParser(grammar)
result = chunk_parser.parse(gram_pos_tag)
print(result)
```

```python
result.draw() # uncomment this line to visualize the parse tree
# Note: The draw method opens a new window to display the parse tree, which may not work in all environments.
# If you are running this in a Jupyter notebook or similar environment, you might want to use `result.pretty_print()` instead.
```

Chinking

```python
import nltk
from nltk.tokenize import word_tokenize

# Input sentence
gram = "It's a dangerous business, Frodo, going out your door."

# Tokenize and POS tag
gram_tokenize = word_tokenize(gram)
gram_pos_tag = nltk.pos_tag(gram_tokenize)

# Define grammar with both chunking and chinking
gar = """
NP: {<DT>?<JJ>*<NN>}   # Chunk noun phrases
}<JJ>{                 # Chink adjectives
"""

# Create the RegexpParser with the grammar
chunk_parsers = nltk.RegexpParser(gar)

# Parse the POS-tagged tokens
result = chunk_parsers.parse(gram_pos_tag)

# Print the result
print(result)
```

NER 


```python
import nltk
nltk.download('maxent_ne_chunker_tab')
nltk.download('maxent_ne_chunker')
nltk.download('words')
```

```python
quote = """
... Men like Schiaparelli watched the red planet—it is odd, by-the-bye, that
... for countless centuries Mars has been the star of war—but failed to
... interpret the fluctuating appearances of the markings they mapped so well.
... All that time the Martians must have been getting ready.
...
... During the opposition of 1894 a great light was seen on the illuminated
... part of the disk, first at the Lick Observatory, then by Perrotin of Nice,
... and then by other observers. English readers heard of it first in the
... issue of Nature dated August 2."""
```

```python
import nltk
from nltk.tokenize import word_tokenize
def ner_word(quote):
    words = word_tokenize(quote)
    tags = nltk.pos_tag(words)
    tree = nltk.ne_chunk(tags)
    print(tree)
    return set(
        "".join(i[0] for i in t)
        for t in tree
        if hasattr(t, 'label') and t.label() == 'NE'
    )
```

```python
print(ner_word(quote))
```
