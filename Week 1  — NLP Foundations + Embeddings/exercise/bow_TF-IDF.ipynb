Creating Vocabulary
- building a vocabulyr is gathering all the unique words  from the  corpus and counting their occurance

```python
import re
from collections import defaultdict

corpus = [
    "Tokenization is the process of breaking text into words.",
    "Vocabulary is the collection of unique words.",
    "The process of tokenizing is essential in NLP.",
]
def build_vocabulary(corpus):
    vocabulary = defaultdict(int)
    for sentence in corpus:
        words = re.findall(r'\b\w+\b', sentence.lower())
        for word in words:
            vocabulary[word] += 1
    sorted_vocab = sorted(vocabulary.items(), key=lambda x: x[1], reverse=True)
    return sorted_vocab

vocabulary = build_vocabulary(corpus)
print("Vocabulary:", vocabulary)
```

Manually building a vocabulary can be time-consuming, especially for large corpora. Scikit-learn's CountVectorizer automates this process and allows for more flexible text processing as we will see later

# Bag of Words Implementation Using Python (From Scratch)

### Step 1: Preprocessing the Text Data

```python
from collections import defaultdict
import string
# sample text
corpus = [
    "Python is amazing and fun.",
    "Python is not just fun but also powerful.",
    "Learning Python is fun!",
]

def preprocess(corpus):
    # convert to lower case
    text = corpus.lower()
    # remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))
    # tokenize
    token = text.split()
    return token
processed = [preprocess(sent) for sent in corpus]
print("processed:", processed)
```

# Step 2: Build a Vovabulary
- Now, we need to scan through all the documents and build a complete list of unique words, that is our vocabulary.

```python
vocabulary = set()
for sentence in processed:
    vocabulary.update(sentence)
vocabulary = sorted(list(vocabulary))
print("Vocabulary:", vocabulary)
```

# Step 3: Calculate Word Frequencies and Vectorize


```python
def create_bow_vector(sentence, vocabulary):
    vector = [0] * len(vocabulary)
    for word in sentence:
        if word in vocabulary:
            id = vocabulary.index(word)
            vector[id] += 1
    return vector
bow_vectors = []
print("Bag of Words Vector:")
for sentence in processed:
    bow_vector = create_bow_vector(sentence, vocabulary)
    bow_vectors.append(bow_vector)
    print(bow_vector)
```

# Using Scikit-learn’s CountVectorizer
- Building a Bag of Words model manually is good for learning, but for production applications, you will want to use efficient, optimized libraries like Scikit-learn.
- in python function for tokenization we **CountVectorizer** 
- which is imported from **sklearn.feature_extraction.text**

```python
from sklearn.feature_extraction.text import CountVectorizer
corpus = [
    "Python is amazing and fun.",
    "Python is not just fun but also powerful.",
    "Learning Python is fun!",
]
vector = CountVectorizer()
x = vector.fit_transform(corpus)
print("Vocabulary:", vector.get_feature_names_out())
print("BOW Vectors Reparesentation:")
print(x.toarray())
```

# Part 2 — TF-IDF (Term Frequency – Inverse Document Frequency)

- Term Frequency (TF) represents the frequency of a term in a document.
- Inverse Document Frequency (IDF) reduces the impact of commonly occurring words across multiple documents.
- The TF-IDF score is calculated by multiplying the two metrics.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
corpus = [
    "Python is amazing and fun.",
    "Python is not just fun but also powerful.",
    "Learning Python is fun!",
]
tfidf = TfidfVectorizer()
vocab = tfidf.fit_transform(corpus)
print("Vocabulary:", tfidf.get_feature_names_out())
print("TF-IDF Vector Representation:")
print(vocab.toarray())
```

# BOW and TF-IDF
## step 1: Text Preprocessing
before bow and TF-IDF , we clean and Normalize the text
- ## Steps:
    - LowerCasing 
    - Tokenazation
    - Removal of  Stop words
    - Lemmatize/stem
    - Keep only Alphabets

```python
from nltk import (word_tokenize, WordNetLemmatizer)
from nltk.corpus import stopwords
stopwords = set(stopwords.words("english"))
word_lem = WordNetLemmatizer()

def process(text):
    token = word_tokenize(text.lower())
    token = [word for word in token if word.isalpha()]
    token = [word for word in token if word not in stopwords]
    token = [word_lem.lemmatize(word) for word in token]
    return " ".join(token)
docs = ["Cats are running faster than dogs!", "The dog ran quickly."]
processed_text = [process(doc)  for doc in docs ]
print("preprocessed text:", processed_text)
```

# 2. BoW and TF-IDF after Preprocessing

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
# Part 1 — Bag of Words (BoW)
bow_vector = CountVectorizer()
bow_matrix = bow_vector.fit_transform(processed_text)
print("Vocabulary:", bow_vector.get_feature_names_out())
print("BoW Vector Representation:")
print(bow_matrix.toarray())

# Part 2 — TF-IDF (Term Frequency – Inverse Document Frequency)
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(processed_text)
print("Vocabulary:", tfidf.get_feature_names_out())
print("TF-IDF Vector Representation:")
print(tfidf_matrix.toarray())
```
